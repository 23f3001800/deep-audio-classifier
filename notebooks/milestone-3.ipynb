```python
import torch
import torchaudio.transforms as T

# 1. Create fake raw audio: 1 channel, 16000 samples (1 second)
raw_audio = torch.rand(1, 16000) 

# 2. Set up the Mel-Spectrogram converter with your exact settings
mel_spectrogram_transform = T.MelSpectrogram(
    sample_rate=16000,
    n_fft=400,
    hop_length=160,
    n_mels=64
)

# 3. Pass the audio through the converter
spectrogram_tensor = mel_spectrogram_transform(raw_audio)

# 4. Check the shape
print(spectrogram_tensor.shape) 
# Output: torch.Size([1, 64, 101])
```

### Convolution Output Formula
$$
H_{out} =
\left\lfloor
\frac{H + 2P - K}{S}
\right\rfloor + 1
$$

$$
H_{out} =
\left\lfloor
\frac{64 + 2(1) - 3}{1}
\right\rfloor + 1
$$

```python
import torch
import torch.nn as nn

x = torch.randn(1, 1, 64, 64)

# Define layers
conv = nn.Conv2d(in_channels=1, out_channels=8, kernel_size=3, stride=1, padding=1)
pool = nn.MaxPool2d(kernel_size=2, stride=2)

# Forward pass
x_conv = conv(x)
x_pool = pool(x_conv)

print("After Conv2d shape:", x_conv.shape)
print("After MaxPool2d shape:", x_pool.shape)
```

```python
import torch
import torch.nn as nn

# 1. Create input tensor (Batch, Channels, Height, Width)
input_tensor = torch.randn(1, 1, 64, 64)

# 2. Define layers
conv = nn.Conv2d(in_channels=1, out_channels=16, kernel_size=3, stride=1, padding=1)
pool = nn.MaxPool2d(kernel_size=2, stride=2)

# 3. Pass through layers
out_conv = conv(input_tensor)
print(f"After Conv: {out_conv.shape}") # torch.Size([1, 16, 64, 64])

out_pool = pool(out_conv)
print(f"After Pool: {out_pool.shape}") # torch.Size([1, 16, 32, 32])
```
